```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report
```

```python
df = pd.read_csv("creditcard.csv")
df
```

```python
features = df.drop(columns=['Class'])
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(features)
```

```python
iso_forest = IsolationForest(n_estimators=100, contamination=0.002, random_state=42)
df['anomaly'] = iso_forest.fit_predict(X_scaled)
```

```python
df['anomaly'] = df['anomaly'].map({1: 0, -1: 1})  
print("Predicted Fraudulent Transactions:", df['anomaly'].sum())
print("Actual Fraudulent Transactions:", df['Class'].sum())
```

```python
cm = confusion_matrix(df['Class'], df['anomaly'])
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```

```python
print(classification_report(df['Class'], df['anomaly']))
```
